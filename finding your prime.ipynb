```python
import numpy as np

primes = []

n = np.linspace(1, 99, 99, dtype=int)

N = np.max(n)

for i in range(N):
    t = 0
    for k in range(N):
        p = n[i] / n[k]
        if p % 1 == 0:
            t = t + 1
        else:
            t = t
    
    if t == 2:
        primes = np.append(primes,n[i]) 
        
print(primes)

#Verified that these are the right numbers
```
