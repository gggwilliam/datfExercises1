```python
import numpy as np

n = np.linspace(1, 100, 100)

N = int(np.max(n))

steps = []

maks = []

for i in range(N): 
    x = n[i]
    count = 0
    max_val = x
    while x != 1:
        if x % 2 == 0:
            x = x/2
        else:
            x = x * 3 + 1 
        count = count + 1
        if x > max_val:
            max_val = x
    steps = np.append(steps, count)
    maks = np.append(maks, max_val)
   
#print(maks)
#print(steps)

max_steps = np.max(steps)

max_start_number = np.argmax(steps) + 1 #Because im starting at 1, and python 0-indexes

print("The number", max_start_number, "uses", int(max_steps), "steps to get to 1")

intmax = int(np.max(maks))

print("Across all starting numbers, the highest integer we encounter is", intmax)
```
