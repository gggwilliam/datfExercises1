```python
import numpy as np

n = np.linspace(1, 100, 100)

N = int(np.max(n))

steps = []


for i in range(N): 
    x = n[i]
    count = 0
    while x != 1:
        if x % 2 == 0:
            x = x/2
        else:
            x = x * 3 + 1 
        count = count + 1
    steps = np.append(steps, count)


#print(steps)

max_steps = np.max(steps)

max_start_number = np.argmax(steps)

print("The number", max_start_number, "uses", int(max_steps), "steps to get to 1")
```
