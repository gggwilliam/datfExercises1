```python
import numpy as np

data = np.genfromtxt("detector_events.txt", delimiter=':', dtype=str)

print(data[3,1])

print(data[3,0][0])

N = len(data)

success = []

for i in range(N):
    min = int(data[i,0][0])
    max = int(data[i,0][2])
    count = 0
    particle = data[i,0][4]
    for k in range(len(data[i,1])):
        if data[i,1][k] == particle:
            count = count + 1
    if min <= count <= max:
        success = np.append(success, 1)
    else:
        success = np.append(success, 0)

print(success)



```
