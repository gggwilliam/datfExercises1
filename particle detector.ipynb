```python
import numpy as np

data = np.genfromtxt("detector_events.txt", delimiter=':', dtype=str)

N = len(data)

success = []

for i in range(N):
    crit = data[i,0].split()
    min, max = crit[0].split('-')
    min = int(min)
    max = int(max)
    particle = crit[1]
    count = 0
    for k in range(len(data[i,1])):
        if data[i,1][k] == particle:
            count = count + 1
        else:
            count = count
    if min <= count <= max:
        success = np.append(success, 1)
    else:
        success = np.append(success, 0)

#print(success)

n_successes = int(np.sum(success))

print("Number of accepted events:", n_successes)

```
